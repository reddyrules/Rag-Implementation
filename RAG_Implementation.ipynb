### Vector Database

```python
from sentence_transformers import SentenceTransformer
```

```python
from langchain_community.vectorstores import FAISS
```

```python
pip install -q langchain-huggingface
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(model_name = "BAAI/bge-small-en-v1.5")
```

```python
vectorstore = FAISS.from_documents(
                         rec_chunk,
                         embedding_model
)
```

```python
retriever = vectorstore.as_retriever()
```

### Chat Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
template = """You are a kind assistent for question answering system for performing tasks based on user query.
                use the following chunks of contextual informatin which is retrieved vector database to answer user query

                if you don't know the answer .... just say i don't have any answer Please don't provide any fabricated information.

                Generate final output using chunks of following contexual information and 
                please restrict information to maximum 10 lines.
                Question : {Question}
                Context : {context}
                Answer :      """
```

```python
prompt = ChatPromptTemplate.from_template(template)
```

```python
prompt
```

### String Output parser

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
str_parser = StrOutputParser()
```

### Chain

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
chain = {"context" : retriever,"Question" : RunnablePassthrough()} | prompt | llm_model | str_parser
```

```python
result = chain.invoke("How USA is Supporting Ukrain in war with Russia")
result
```

```python
result = chain.invoke("What Action USA is taking to Address domestic issues like rise in price of daily goods like prices")
result
```
